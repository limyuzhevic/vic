### Mathematical Background

The single neuron learning experiment is based on the following principles:

#### 1. Leaky Integrate-and-Fire Neuron Model

The neuron membrane potential follows the differential equation:
```
dV/dt = (-V + I_syn)/τ_m
```

Where:
- V = membrane potential (mV)
- I_syn = synaptic current (normalized units)
- τ_m = membrane time constant (default: 20 ms)

When V reaches threshold (typically -50 mV), the neuron emits a spike and resets to -75 mV.

#### 2. Hebbian Learning Rule

Synaptic weights are updated according to:
```
Δw = η × x × y
```

Where:
- Δw = change in synaptic weight
- η = learning rate (typically 0.001)
- x = pre-synaptic activity (input neuron)
- y = post-synaptic activity (main neuron)

#### 3. Learning Metrics

We track several key metrics to quantify learning:

**Firing Rate**: Number of spikes per unit time
**Response Adaptability**: How much response changes over time
**Synaptic Efficacy**: Average strength of neural connections
**Learning Efficiency**: Ratio of learned improvement to training time

### Experimental Design

The experiment follows a structured approach:

#### Phase 1: Baseline Measurement
- Present 10 weak stimuli to establish initial response level
- Measure spontaneous firing rate and baseline synaptic weights

#### Phase 2: Learning Trials
- Present 100 stimuli with varying strengths (weak → moderate → strong)
- Track neural response changes over time
- Record spike times and synaptic modifications

#### Phase 3: Analysis
- Compare initial and final responses
- Determine if learning is statistically significant
- Classify learning type (facilitation, depression, or no learning)
